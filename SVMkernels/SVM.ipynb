```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

x=np.linspace(0,10,100)
y=np.sqrt(x)
y=np.hstack((y,-y))
x=np.hstack((x,-x))
```

```python
x1=np.linspace(0,10,100)
y1=np.sqrt(x1)
y1=np.hstack((y1,-y1))
x1=np.hstack((x1,-x1))
```

```python
plt.scatter(x,y)
plt.scatter(x1,y1)
```

```python
import pandas as pd
df=pd.DataFrame(np.vstack((x,y)).T,columns=['x','y'])
df['Y']=0
df2=pd.DataFrame(np.vstack((x1,y1)).T,columns=['x','y'])
df2['Y']=1
df = pd.concat([df, df2], ignore_index=True)
df.head(5)
```

```python
x=df.iloc[:,0:2]
y=df.Y
```

```python
y
```

```python
from sklearn.svm import SVC
classifier = SVC(kernel='linear')
classifier.fit(x,y)
```

```python
from sklearn.metrics import accuracy_score
y_pred = classifier.predict(x)
accuracy_score(y,y_pred)
```
